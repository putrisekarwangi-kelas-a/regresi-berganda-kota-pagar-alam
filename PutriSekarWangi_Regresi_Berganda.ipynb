```python
#Import Library
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Membuat DataFrame dari data yang diberikan
data = {
'Partisipan': np.arange(1, 11),
'PMakan': [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
'PTranspor': [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
'BB': [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]}

df = pd.DataFrame(data)

# Memisahkan variabel independen dan dependen
X = df[['PMakan', 'PTranspor']]
y = df['BB']

X = sm.add_constant(X) # Menambahkan kolom konstanta untuk intercept

model = sm.OLS(y, X) # Membangun model regresi
results = model.fit()

print(results.summary()) # Menampilkan hasil
```

## VISUAL 3 DIMENSI

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

data = {
    "Partisipan": np.arange(1, 11),
    "Peng Makanan": [50, 40, 60, 55, 45, 65, 70, 75, 80, 90],
    "Peng Transportasi": [20, 25, 30, 35, 40, 50, 55, 60, 65, 70],
    "Berat Badan": [60, 55, 65, 70, 62, 75, 80, 85, 90, 95]
}
df = pd.DataFrame(data)

# Variabel independen
X = df[["Peng Makanan", "Peng Transportasi"]]
# Variabel dependen
y = df["Berat Badan"]

# 4. MENAMBAHKAN INTERCEPT
X_model = sm.add_constant(X)

# 5. ESTIMASI MODEL OLS
model = sm.OLS(y, X_model)
results = model.fit()

# 6. MENAMPILKAN RINGKASAN HASIL REGRESI
print("\n")
print("=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA")
print("=" * 70)
print(results.summary())

# 7. MENGAMBIL KOEFISIEN REGRESI
intercept = results.params["const"]
coef_makanan = results.params["Peng Makanan"]
coef_transportasi = results.params["Peng Transportasi"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# 8. MEMBUAT PERSAMAAN REGRESI
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_makanan:.3f}) X1"
    f" + ({coef_transportasi:.3f}) X2"
)
print("\n")
print("=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# 9. MENGHITUNG NILAI PREDIKSI
df["Prediksi"] = results.predict(X_model)
print("\n")
print("=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(
    df[["Partisipan", "Peng Makanan", "Peng Transportasi", "Berat Badan", "Prediksi"]].round(2)
)

# 10. MEMBUAT GRID UNTUK BIDANG REGRESI
x1 = np.linspace(
    df["Peng Makanan"].min(),
    df["Peng Makanan"].max(),
    40
)
x2 = np.linspace(
    df["Peng Transportasi"].min(),
    df["Peng Transportasi"].max(),
    40
)
X1, X2 = np.meshgrid(x1, x2)

# 11. MENGHITUNG NILAI Y PADA BIDANG REGRESI
Y = (intercept + coef_makanan * X1 + coef_transportasi * X2)

# 12. MEMBUAT FIGURE 3D
fig = go.Figure()

# 13. MENAMPILKAN DATA AKTUAL
fig.add_trace(
    go.Scatter3d(
        x=df["Peng Makanan"],
        y=df["Peng Transportasi"],
        z=df["Berat Badan"],
        mode="markers",
        marker=dict(
            size=7,
            color="red",
            opacity=0.9
        ),
        name="Data Aktual",
        text=[
            f"Partisipan {p}"
            for p in df["Partisipan"]
        ],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Peng. Makanan: %{x}<br>"
            "Peng. Transportasi: %{y}<br>"
            "Berat Badan: %{z}"
            "<extra></extra>"
        )
    )
)

# 14. MENAMPILKAN BIDANG REGRESI
fig.add_trace(
    go.Surface(
        x=X1, y=X2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Regression Plane",
        hovertemplate=(
            "Peng. Makanan: %{x:.2f}<br>"
            "Peng. Transportasi: %{y:.2f}<br>"
            "Prediksi Y: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

# 15. MENGATUR TAMPILAN GRAFIK
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D"
            "<br>"
            f"<sup>{persamaan} | "
            f"R2 = {r_squared:.3f}</sup>"
        ), "x": 0.5
    },
    scene={
        "xaxis": {"title": "Pengeluaran Makanan (X1)"},
        "yaxis": {"title": "Pengeluaran Transportasi (X2)"},
        "zaxis": {"title": "Berat Badan (Y)"}
    },
    width=1000,
    height=750,
    margin={
        "l": 0, "r": 0, "b": 0, "t": 100
    },
    legend={"x": 0.02, "y": 0.98}
)

# 16. MENAMPILKAN GRAFIK
fig.show()
```

# REGRESI BERGANDA

## Indikator Variabel
Y = IPM

X1 = bm_pendidikan, X2 = bm_kesehatan, X3 = bm_infrastruktur

Kota = Kota Pagar Alam

### Import & Ambil Data Kota Pagar Alam

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

# 1. LOAD DATA & FILTER 1 KOTA
df_all = pd.read_csv("dataset_fiskal_97_hipoFull.csv")
df = df_all[df_all["Kota"] == "Kota Pagar Alam"].reset_index(drop=True)

# 2. PILIH VARIABEL & RAPIKAN NAMA
df = df[["Tahun", "IPM", "bm_pendidikan", "bm_kesehatan", "bm_infrastruktur"]]
df.columns = ["Tahun", "IPM", "Belanja Pendidikan", "Belanja Kesehatan", "Belanja Infrastruktur"]

print(df.shape)   # harusnya (11, 5)
df
```

### Regresi berganda

```python
# Variabel independen (X) dan dependen (Y)
X = df[["Belanja Pendidikan", "Belanja Kesehatan", "Belanja Infrastruktur"]]
y = df["IPM"]

# Menambahkan intercept
X_model = sm.add_constant(X)

# Estimasi OLS
model = sm.OLS(y, X_model)
results = model.fit()

print("=" * 70)
print(" HASIL REGRESI LINEAR BERGANDA - KOTA PAGAR ALAM")
print("=" * 70)
print(results.summary())

# Ambil koefisien
intercept = results.params["const"]
b1 = results.params["Belanja Pendidikan"]
b2 = results.params["Belanja Kesehatan"]
b3 = results.params["Belanja Infrastruktur"]
r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# Persamaan regresi
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({b1:.3f}) X1"
    f" + ({b2:.3f}) X2"
    f" + ({b3:.3f}) X3"
)

print("\n" + "=" * 70)
print(" PERSAMAAN REGRESI")
print("=" * 70)
print(persamaan)
print(f"\nR2 = {r_squared:.4f}")
print(f"Adjusted R2 = {adjusted_r_squared:.4f}")

# Nilai prediksi
df["Prediksi"] = results.predict(X_model)
print("\n" + "=" * 70)
print(" DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(df.round(2))
```

### Visualisasi 3D

```python
# Grid untuk bidang regresi (X1 dan X2)
x1 = np.linspace(df["Belanja Pendidikan"].min(), df["Belanja Pendidikan"].max(), 40)
x2 = np.linspace(df["Belanja Kesehatan"].min(), df["Belanja Kesehatan"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)

# X3 dikunci di rata-rata
x3_rata = df["Belanja Infrastruktur"].mean()

# Nilai Y pada bidang regresi
Y = intercept + b1 * X1 + b2 * X2 + b3 * x3_rata

fig = go.Figure()

# Data aktual
fig.add_trace(
    go.Scatter3d(
        x=df["Belanja Pendidikan"],
        y=df["Belanja Kesehatan"],
        z=df["IPM"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Tahun {t}" for t in df["Tahun"]],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Belanja Pendidikan: %{x}<br>"
            "Belanja Kesehatan: %{y}<br>"
            "IPM: %{z}"
            "<extra></extra>"
        )
    )
)

# Bidang regresi
fig.add_trace(
    go.Surface(
        x=X1, y=X2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Bidang Regresi",
        hovertemplate=(
            "Belanja Pendidikan: %{x:.2f}<br>"
            "Belanja Kesehatan: %{y:.2f}<br>"
            "Prediksi IPM: %{z:.2f}"
            "<extra></extra>"
        )
    )
)

fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D - KOTA PAGAR ALAM"
            "<br>"
            f"<sup>{persamaan} | R2 = {r_squared:.3f}<br>"
            f"(X3 dikunci di rata-rata = {x3_rata:.1f})</sup>"
        ), "x": 0.5
    },
    scene={
        "xaxis": {"title": "Belanja Pendidikan (X1)"},
        "yaxis": {"title": "Belanja Kesehatan (X2)"},
        "zaxis": {"title": "IPM (Y)"}
    },
    width=1000,
    height=750,
    margin={"l": 0, "r": 0, "b": 0, "t": 120},
    legend={"x": 0.02, "y": 0.98}
)

fig.show()
```

### Aktual VS Prediksi per tahunn

```python
fig2 = go.Figure()
fig2.add_trace(go.Scatter(x=df["Tahun"], y=df["IPM"], mode="lines+markers", name="IPM Aktual"))
fig2.add_trace(go.Scatter(x=df["Tahun"], y=df["Prediksi"], mode="lines+markers", name="IPM Prediksi"))
fig2.update_layout(
    title="IPM Aktual vs Prediksi - Kota Pagar Alam",
    xaxis_title="Tahun",
    yaxis_title="IPM",
    template="simple_white"
)
fig2.show()
```

##### Interpretasi
Persamaan regresi

Ŷ = 84,858 + (−0,003) X1 + (−0,022) X2 + (−0,064) X3

1. Uji F (simultan)

F-hitung = 2,288, Prob (F-statistic) = 0,166, lebih besar dari 0,05.
Artinya secara bersama-sama, belanja pendidikan, kesehatan, dan infrastruktur belum berpengaruh signifikan terhadap IPM Kota Pagar Alam.

2. Uji t (parsial)

- Variabel Belanja Pendidikan (X1) ; Koefisien −0,003 ; p-value 0,924	;Kesimpulan: Tidak signifikan
- Variabel Belanja Kesehatan (X2) ; Koefisien	−0,022 ; p-value 0,379 ; Kesimpulan:	Tidak signifikan
- Variabel Belanja Infrastruktur (X3) ; Koefisien	−0,064 ; p-value	0,041	; Kesimpulan: Signifikan (negatif)

X1 dan X2 tidak berpengaruh signifikan terhadap IPM.
X3 signifikan pada taraf 5%. Setiap kenaikan belanja infrastruktur 1 satuan, IPM turun sekitar 0,064 poin (variabel lain tetap).
Arah negatif ini berlawanan dengan teori, karena harusnya belanja makin besar, IPM makin baik. Bisa dijelaskan sebagai efek jangka panjang (belanja infrastruktur baru terasa ke IPM beberapa tahun kemudian), atau karena datanya cuma 11 tahun sehingga hasilnya belum stabil. Jangan disimpulkan “infrastruktur merusak IPM” ya.
Karena F-test-nya tidak signifikan, signifikansi X3 ini sebaiknya diperlakukan hati-hati.

3. Koefisien determinasi

R² = 0,495, artinya sekitar 49,5% variasi IPM dijelaskan oleh ketiga belanja tersebut, sisanya (50,5%) oleh faktor lain di luar model.
Adjusted R² = 0,279. Angka ini jauh lebih kecil karena sudah dikoreksi jumlah variabel terhadap jumlah data yang sedikit, jadi kemampuan model sebenarnya lebih lemah dari kelihatannya.

4. Intercept

Konstanta = 84,858 (signifikan). Ini perkiraan IPM kalau semua belanja bernilai 0, jadi cukup disebut sebagai titik awal model, tidak perlu dimaknai secara nyata.

5. Visualisasi 3D

Bidang regresinya hampir datar dan sedikit miring turun, titik merah (data aktual) tersebar jauh dari bidang. Ini selaras dengan R² yang hanya sedang dan X1, X2 yang tidak signifikan.
Di grafik aktual vs prediksi, garis prediksi mengikuti pola kasar IPM tapi meleset di beberapa tahun, misalnya IPM 2015 yang tinggi (84,14) tidak tertangkap dengan baik.

Kesimpulan singkat

Hasil regresi linear berganda untuk Kota Pagar Alam (2014–2024) menunjukkan bahwa belanja pendidikan, kesehatan, dan infrastruktur secara simultan tidak berpengaruh signifikan terhadap IPM (Prob F = 0,166). Secara parsial hanya belanja infrastruktur yang signifikan dengan arah negatif (p = 0,041), sedangkan belanja pendidikan dan kesehatan tidak signifikan. Model menjelaskan 49,5% variasi IPM (Adjusted R² = 27,9%). Keterbatasan jumlah observasi (11 data) membuat hasil ini perlu ditafsirkan hati-hati.
